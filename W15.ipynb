```python
from sklearn import tree
from sklearn.metrics import confusion_matrix, accuracy_score
from matplotlib import pyplot as plt
```

```python
#학습용 데이터
X_train = [[1,1],[1,1],[1,1],[1,0],[2,1],
           [1,21],[2,1],[2,1],[2,0],[2,0]]
y_train = [1,1,1,1,0,0,0,1,1,1]

#테스트용 데이터
X_test = [[1,1],[1,0],[2,1],[2,0]]
y_test = [1,1,0,1]
```

```python
#C5.0 모형 설정
clf = tree.DecisionTreeClassifier(criterion='entropy')
```

```python
# 학습
clf = clf.fit(X_train,y_train)
# 그래프 출력 영역(가로 세로, 단위는 인치
plt.figure(figsize=(6,6))

#학습된 모형 출력
tree.plot_tree(clf)
plt.show()
```

```python
# 분류예측
y_pred = clf.predict(X_test)
y_pred
```

```python
# 평가 - 혼동행렬 출력
print(confusion_matrix(y_test,y_pred));
```

```python
# 정확도 출력
print("정확도: ", accuracy_score(y_test,y_pred));
```

```python
# 입력 항목명
iris = load_iris()
iris
```

```python
# 입력 항목명
print(iris.feature_names)
# 목표 클래스의 유형(이름)
print(iris.target_names)
```

```python
# 입력 데이터 세트
x = iris.data
# 타겟 데이터 세트
y = iris.target
```

```python
# 입력 데이터 세트의 행과 열의 크기
print(x.shape)
# 입력 데이터 세트 출력
print(x)
```

```python
# 목표 데이터 세트의 행과 열의 크기
print(y.shape)
# 타겟 데이터 세트 출력
print(y)
```

```python
#데이터 세트 분리 : 학습용 입력값, 테스트용 입력, 학습용 타겟, 테스트용 타겟
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=1234)
```

```python
# 학습용 입력 데이터 세트의 행과 열의 크기 출력
print(X_train.shape)
# 학습용 입력값의 출력
print(X_train)
```

```python
# 학습용 목표 데이터 세트의 행과 열의 크기 출력
print(y_train.shape)
# 학습용 목표값의 출력
print(y_train)
```

```python
# 테스트용 입력 데이터 세트의 행과 열의 크기 출력
print(X_test.shape)
# 테스트용 입력값의 출력
print(X_test)
```

```python
#C5.0 모형 설정
clf = tree.DecisionTreeClassifier(criterion='entropy')
```

```python
# 학습
clf = clf.fit(X_train,y_train)
# 그래프 출력 영역(가로 세로, 단위는 인치
plt.figure(figsize=(15,12))

#학습된 모형 출력
tree.plot_tree(clf)
plt.show()
```

```python
#분류예측
y_pred = clf.predict(X_test)
y_pred
```

```python
# 평가: 혼동행렬 출력
print(confusion_matrix(y_test,y_pred))
```

```python
# 정확도 출력
print("정확도: ", accuracy_score(y_test,y_pred));
```
