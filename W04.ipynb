```python
#주성분 분석
from sklearn.decomposition import PCA
#데이터 표준화
from sklearn.preprocessing import StandardScaler

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 원데이터
X = np.array([[50,73], [65,75], [75,80], [80,82], [95,85]])
X

# 원데이터의 1열(X1)과 2열(X2)의 산포도
plt.scatter(X[:,0], X[:,1])
plt.xlabel('X1')
plt.ylabel('X2')
plt.xlim(0,100)
plt.ylim(0,100)
# plt.show()

#열별 평균
print(np.mean(X, axis=0))

#열별 분산
print(np.var(X, axis=0))

#데이터 표준화
#표준화 스케일러 객체 생성
scalerX = StandardScaler()
# 표준화 스케일러를 x의 각 열에 기준을 맞춤
scalerX.fit(X)
# x의 각 열을 표준화로 변환
X_std = scalerX.transform(X)

# 데이터 프레임으로 변환 및 열 이름 설정
print(X_std)


# 표준화된 데이터의 1열(X1)과 2열(X2)의 산포도
plt.scatter(X_std[:,0], X_std[:,1])
plt.xlabel('X1')
plt.ylabel('X2')
plt.xlim(-3,3)
plt.ylim(-3,3)


#열별 평균
print(np.mean(X_std, axis=0))

#모공분산행렬
print(np.cov(X_std[:,0],X_std[:,1],ddof = 0))

# 주성분의 수 설정
pca = PCA(n_components = 2)

# 주성분 분석
pca.fit(X_std)

# 주성분 분산
print(pca.explained_variance_)
# 주성분 분산비율
print(pca.explained_variance_ratio_)

# 주성분으로 데이터 변환
Z = pca.fit_transform(X_std)
Z

# 주성분 좌표계의 분포
plt.scatter(Z[:,0], Z[:,1])
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.xlim(-2.5,2.5)
plt.ylim(-2.5,2.5)

#주성분로딩(고유벡터)
loadings = pca.components_
print(loadings)


# 시각화 패스
```
