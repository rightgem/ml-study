```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
#iris 데이터 세트
from sklearn import datasets

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#iris 데이터 세트 읽기
data = datasets.load_iris()
data
```

```python
#data 부분에 있는 feature_names을 열 이름으로 하는 데이터 프레임
df = pd.DataFrame(data['data'], columns=data['feature_names'])
#target 부분의 데이터를 데이터 프레임에 열로추가
df['target'] = data['target']
# 데이터 프레임 앞부분 출력
df.head()
```

```python
df.tail()
```

```python
#그래프 출력
plt.figure(figsize=(8,8)) # 8 인치 장표 출력
ax = sns.pairplot(df, hue='target') # 두 변수 조합의 target
plt.show()
```

```python
# data 부분
X_df = df.iloc[:, [0,1,2,3]]
# 열별 평균
print(X_df.mean(axis = 0))
# 열별 분산
print(X_df.var(axis = 0))
```

```python
# 데이터 표준화
scalerX = StandardScaler()
scalerX.fit(data.data)
X_std = scalerX.transform(data.data)
X_std
```

```python
# 주성분 분석 객체 생성
pca = PCA()
# 주성분 분석
pca.fit(X_std)

# 주성분 분산비율
print(pca.explained_variance_ratio_)
# 주성분 분산
print(pca.explained_variance_)

plt.plot(pca.explained_variance_, 'o-')
plt.title('Scree Plot')
plt.xlabel('Principal Component')
plt.ylabel('variance')
```

```python
# 주성분으로 데이터 변환
Z = pca.fit_transform(X_std)
# 데이터프레임 변환
Z_df = pd.DataFrame(data = Z, columns = ['PC1', 'PC2', 'PC3', 'PC4'])
Z_df.head()
```

```python
#target 열 추가
Z_df['target'] = data.target
# 그래프 출력
plt.figure(figsize=(8,8))
ax = sns.pairplot(Z_df, hue='target')
plt.show()
```

```python
# 주성분 로딩(고유 벡터)
loadings = pca.components_
print(loadings)
```
